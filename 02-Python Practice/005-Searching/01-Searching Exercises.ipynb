**1. Linear Search - Python**

Input: arr[] = [10, 50, 30, 70, 80, 20, 90, 40], x = 30<br>
Output : 2

```python
l = [10, 50, 30, 70, 80, 20, 90, 40]
print(l)

x = int(input("Enter the Element you want to Search: "))

for i in range(len(l)):
    if l[i] == x:
        print(i) 
```

**2. Binary Search**

```python
def binary(l,x):
    low  = 0
    high = len(l)-1

    while low <= high:
        mid = (low + high) // 2

        if l[mid] == x:
            return mid
        elif l[mid] > x:
            high = mid - 1
        else:
            low = mid + 1
    return -1

l = [10, 50, 30, 70, 80, 20, 90, 40]
l.sort()
print(l)

x = int(input("Enter the Element you want to Search: "))

r = binary(l,x)

if r != -1:
    print(f"Element found at index {r}")
else:
    print("Element not found")
```
