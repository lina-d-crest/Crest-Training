**1. Sort a String**

Given a string consisting of lowercase letters, arrange all its letters in ascending order. <br>

Examples:<br>

Input:<br>
S = "edcab"<br>
Output: "abcde"<br>

Input:<br>
S = "xzy"<br>
Output: "xyz"<br>

```python
def sort(s):
    return ''.join(sorted(s))
```

```python
s = "edcab"
print(sort(s))
```

```python
s = "xzy"
print(sort(s))
```

**2. Bubble Sort**

```python
def bubble(l):
    
    for i in range(len(l)):
        for j in range(0,len(l) - i - 1):
            if l[j] > l[j+1]:
                l[j], l[j+1] = l[j+1], l[j]
                
    return l
```

```python
arr = [6,6,2,12,5,1,3]
print("Unsorted list is:")
print(arr)

bubble(arr)

print("Sorted list is:")
print(arr)
```

**3. Insertion Sort**

```python
def insertion_sort(arr):
    
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

    return arr
```

```python
arr = [6,6,2,12,5,1,3]
print("Unsorted list is:")
print(arr)

insertion_sort(arr)

print("Sorted list is:")
print(arr)
```

**4. Selection Sort**

```python
def selection_sort(array):

    size = len(arr)
    for ind in range(size):
        min_index = ind

        for j in range(ind + 1, size):
            if array[j] < array[min_index]:
                min_index = j

        array[ind], array[min_index] = array[min_index], array[ind]
```

```python
arr = [6,6,2,12,5,1,3]
print("Unsorted list is:")
print(arr)

selection_sort(arr)

print("Sorted list is:")
print(arr)
```

**5. Quick Sort**

```python
def quick_sort(a, low, high):
    if low < high:
        pi = partition(a, low, high)
        quick_sort(a, low, pi - 1)
        quick_sort(a, pi + 1, high)

def partition(a, low, high):
    pivot = a[high]
    i = low - 1

    for j in range(low, high):
        if a[j] <= pivot:
            i += 1
            a[i], a[j] = a[j], a[i]

    a[i + 1], a[high] = a[high], a[i + 1]
    return i + 1
```

```python
arr = [6,6,2,12,5,1,3]
print("Unsorted list is:")
print(arr)

quick_sort(arr,0,len(arr)-1)

print("Sorted list is:")
print(arr)
```
