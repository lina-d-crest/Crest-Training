**1. Header Information and Parsing**

Write a Python program to send a request to a web page, and print the header information. Also parse these values and print key-value pairs holding various information.

```python
import requests
```

```python
r = requests.get('https://api.github.com/')
response = r.headers

print("Headers information of the said response:")
print(response)
```

```python
print("\nVarious Key-value pairs information of the said resource and request:\n")

print("Date:", response.get('date'))
print("Server:", response.get('server'))

# HTTP status is not in headers; access it like this:
print("Status:", r.status_code, r.reason)

print("Cache-Control:", response.get('cache-control'))
print("Vary:", response.get('vary'))
print("X-GitHub-Media-Type:", response.get('x-github-media-type'))
print("Access-Control-Expose-Headers:", response.get('access-control-expose-headers'))
print("Strict-Transport-Security:", response.get('strict-transport-security'))
print("X-Content-Type-Options:", response.get('x-content-type-options'))
print("X-XSS-Protection:", response.get('x-xss-protection'))
print("Referrer-Policy:", response.get('referrer-policy'))
print("Content-Security-Policy:", response.get('content-security-policy'))
print("Content-Encoding:", response.get('content-encoding'))
print("X-RateLimit-Remaining:", response.get('x-ratelimit-remaining'))
print("X-RateLimit-Reset:", response.get('x-ratelimit-reset'))
print("X-RateLimit-Used:", response.get('x-ratelimit-used'))
print("Accept-Ranges:", response.get('accept-ranges'))
print("X-GitHub-Request-Id:", response.get('x-github-request-id'))
```
