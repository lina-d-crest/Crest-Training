**1. Random Value Generation**

Write a Python program to generate a random color hex, a random alphabetical string, random value between two integers (inclusive) and a random multiple of 7 between 0 and 70.

Use random.randint()

```python
import random
import string
```

```python
print("Generate a random color hex:")
print("#{:06x}".format(random.randint(0, 0xFFFFFF)))
print("\nGenerate a random alphabetical string:")

max_length  = 255
s = ""

for i in range(random.randint(1,max_length )):
    s +=random.choice(string.ascii_letters)

print(s)
print("\nGenerate a random value between two integers, inclusive:")
print(random.randint(0, 10))
print(random.randint(-7, 7))
print(random.randint(1, 1))
print("\nGenerate a random multiple of 7 between 0 and 70:")
print(random.randint(0, 10) * 7)
```

<br>**2. Shallow Copy of List**

Write a Python program to create a shallow copy of a given list. Use copy.copy

```python
import copy
```

```python
nums_x = [1, [2, 3, 4]]
print("Original list: ", nums_x)
nums_y = copy.copy(nums_x)
print("\nCopy of the said list:",nums_y)
```

```python
print("\nChange the value of an element of the original list:")
nums_x[1][1] = 10
print(nums_x)
print("\nSecond list:")
print(nums_y)
```

```python
nums =  [[1], [2]]
nums_copy = copy.copy(nums)
print("\nOriginal list:")
print(nums)
print("\nCopy of the said list:")
print(nums_copy)
```

```python
print("\nChange the value of an element of the original list:")
nums[0][0] = 0
print("\nFirst list:")
print(nums)
print("\nSecond list:")
print(nums_copy)
```
