**1. Interval Intersection**

Write a Python program to find the overlapping period from two given two time intervals using Pendulum module.

```python
pip install pendulum
```

```python
import pendulum
```

```python
is1 = "2020-03-15T08:00:00"
ie1 = "2020-03-15T12:00:00"

is2 = "2020-03-15T10:30:00"
ie2 = "2020-03-15T14:00:00"
```

```python
ist1 = pendulum.parse(is1)
iet1 = pendulum.parse(ie1)

ist2 = pendulum.parse(is2)
iet2 = pendulum.parse(ie2)

os = max(ist1,ist2)
oe = min(iet1,iet2)

if os <= oe:
    print(f"There is an overlapping period from {os} to {oe}.")
else:
    print("There is no overlapping period.")
```

Write a Python program to convert the current date and time to a Unix timestamp using Pendulum module.
