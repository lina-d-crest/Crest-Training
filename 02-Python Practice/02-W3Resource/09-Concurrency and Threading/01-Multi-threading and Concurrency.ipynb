**1. Multithreaded Factorial Calculator**

Write a Python program to calculate the factorial of a number using multiple threads.

```python
import threading
```

```python
def fact(n):

    r = 1
    for i in range(1,n+1):
        r *= i
        return r
```

```python
def cal(n):
    print(f"Calculating factorial of {n} in thread {threading.current_thread().name}")
    result = fact(n)
    print(f"Factorial of {n} is {result} in thread {threading.current_thread().name}")
```

```python
n = 12

t1 = threading.Thread(target=cal, args=(n,))
t2 = threading.Thread(target=cal, args=(n,))

t1.start()
t2.start()

t1.join()
t2.join()
```

<br>**2. Concurrent HTTP Requests**

Write a Python program that performs concurrent HTTP requests using threads.

```python
import requests
```

```python
def fun(url):
    
    r = requests.get(url)
    print(f"Response from {url}: {r.status_code}")
```

```python
urls = [
    "https://www.example.com",
    "https://www.google.com",
    "https://www.wikipedia.org",
    "https://www.python.org"
]
```

```python
threads = []
for url in urls:
    thread = threading.Thread(target=fun, args=(url,))
    thread.start()
    threads.append(thread)

for thread in threads:
    thread.join()
```
