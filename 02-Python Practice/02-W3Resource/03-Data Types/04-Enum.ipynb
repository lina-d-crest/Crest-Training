**1. Create an Enum object and show a member name and value**

Write a Python program to create an Enum object and display a member name and value.

```python
import enum
```

```python
class Country(enum.Enum):
    Afghanistan = 93
    Albania = 355
    Algeria = 213
    Andorra = 376
    Angola = 244
    Antarctica = 672
    India = 355
    USA = 213
```

```python
print("Distinct Enum Members:")
for country in Country:
    print(f"{country.name} = {country.value}")
```

```python
print("All Enum Members (with aliases):")
for name, member in Country.__members__.items():
    print(f"{name} = {member.value}")
```

```python
print('\nMember name: {}'.format(Country.Albania.name))
print('Member value: {}'.format(Country.Albania.value))
```
