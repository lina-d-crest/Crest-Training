 **1. Word Frequency Sorting**

Write a Python program that creates a counter of the words in a sentence and prints the words in ascending and descending order of their frequency.

```python
from collections import Counter
```

```python
def main():

    s = "Red Green Black Black Red red Orange Pink Pink Red White."
    w = s.split()
    wc = Counter(w)

    print("Words in Ascending Order:")

    swa = sorted(wc.items(),key=lambda item: item[1])

    for i,j in swa:
        print(f"{i}: {j}")

    print("\n")

    print("Words in Descending Order:")
    swd = sorted(wc.items(), key=lambda item: item[1], reverse=True)
    
    for i,j in swd:
        print(f"{i}: {j}")
```

```python
main()
```
