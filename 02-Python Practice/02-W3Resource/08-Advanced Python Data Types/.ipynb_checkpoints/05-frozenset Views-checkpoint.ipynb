**1. Frozenset Operations**

Write a Python program that performs common set operations like union, intersection, and difference of two frozensets.

```python
def main():

    x = frozenset([1,2,3,4,5])
    y = frozenset([0,1,3,7,8,10])

    print("Original Frozensets: ")
    print(x)
    print(y)

    ur = x.union(y)
    print("\nUnion of two said Frozensets: ",ur)

    ir = x.intersection(y)
    print("Intersection of two saif Frozensets: ",ir)

    d1 = x.difference(y)
    d2 = y.difference(x)

    print("\nDifference of x-y: ",d1)
    print("Difference of y-x: ",d2)
```

```python
main()
```
