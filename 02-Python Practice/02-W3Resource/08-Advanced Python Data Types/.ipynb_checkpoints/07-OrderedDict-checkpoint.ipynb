**1. Sort OrderedDict**

Write a Python program that sorts the OrderedDict by its keys. Sort the OrderedDict by its values as well.

```python
from collections import OrderedDict
```

```python
d = OrderedDict()
d['Laptop'] = 40
d['Desktop'] = 45
d['Mobile'] = 35

print("OrderedDict contents:")
for item, qty in d.items():
    print(f"Item: {item}, Quantity: {qty}")
```

```python
k = OrderedDict(sorted(d.items(), key=lambda x: x[0]))
v = OrderedDict(sorted(d.items(), key=lambda x: x[1]))

print("Sorted by keys:")
for item, qty in k.items():
    print(f"Item: {item}, Quantity: {qty}")

print("\nSorted by values:")
for item, qty in v.items():
    print(f"Item: {item}, Quantity: {qty}")
```
