**1. String to Bytes Conversion**

Write a Python program to convert a given string to bytes using different encodings like UTF-8, UTF-16, ASCII etc.

```python
def fun(s,e):
    try:
        eb = s.encode(e)
        return eb
    except Exception as e:
        print("Error:",e)
        return None
```

```python
def main():
    try:
        s = "Python Exercises!"

        e = ["utf-8", "utf-16", "ascii"]

        for i in e:
            eb = fun(s,i)
            if eb is not None:
                print(f"Encoding: {i.upper()}")
                print("Original String: ",s)
                print("Encoded Bytes: ",eb)
                print("Decoded String: ",eb.decode(i))
                print("="*40)

    except Exception as e: 
        print("An error occured:",e)
```

```python
if __name__ == "__main__":
    main()
```
