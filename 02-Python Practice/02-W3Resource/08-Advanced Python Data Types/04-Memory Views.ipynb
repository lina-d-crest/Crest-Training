**1. Memory View to Bytes Conversion**

Write a Python function that takes a memory view and converts it to a bytes object.

```python
def fun(m):

    try:
        bd = bytes(m)
        return bd
    except Exception as e:
        print("Error Occured: ",e)
        return None
```
