<br>**2. Age Calculator**

Write a Python program to calculate an age in years,months and days.

```python
from datetime import date
```

```python
def cal(dob):
    
    t = date.today()
    y = t.year - dob.year
    m = t.month - dob.month
    d = t.day - dob.day

    if d < 0:
        m -= 1
        d += (date(t.year, t.month, 1) - date(t.year if t.month > 1 else t.year - 1, t.month - 1 if t.month > 1 else 12, 1)).days
    if m < 0:
        y -= 1
        m += 12
    return y, m, d
```

```python
print()
for dob in [date(2004,10,29), date(2004,5,12)]:
    y, m, d = cal(dob)
    print(f"Age for DOB {dob}: {y} years, {m} months, {d} days")
print()
```
