**1. Linear Search - Python**

Input: arr[] = [10, 50, 30, 70, 80, 20, 90, 40], x = 30<br>
Output : 2

```python
l = [10, 50, 30, 70, 80, 20, 90, 40]
print(l)

x = int(input("Enter the Element you want to Search: "))

for i in range(len(l)):
    if l[i] == x:
        print(i) 
```

**2. Binary Search**

```python
def binary(l,x):
    low  = 0
    high = len(l)-1

    while low <= high:
        mid = (low + high) // 2

        if l[mid] == x:
            return mid
        elif l[mid] > x:
            high = mid - 1
        else:
            low = mid + 1
    return -1

l = [10, 50, 30, 70, 80, 20, 90, 40]
l.sort()
print(l)

x = int(input("Enter the Element you want to Search: "))

r = binary(l,x)

if r != -1:
    print(f"Element found at index {r}")
else:
    print("Element not found")
```

**3. Anagram Substring Search**

1) Input:  txt[] = "BACDGABCDA"  pat[] = "ABCD"<br>
   Output:  <br> Found at Index 0<br>
             Found at Index 5<br>
             Found at Index 6<br>
2) Input: txt[] =  "AAABABAA" pat[] = "AABA"<br>
   Output:  <br> Found at Index 0<br>
             Found at Index 1<br>
             Found at Index 4<br>

```python
MAX = 256

def compare(a1,a2):
    for i in range(MAX):
        if a1[i] != a2[i]:
            return False
    return True

def search(pat,txt):
    M = len(pat)
    N = len(txt)

    cP = [0] * MAX
    cT = [0] * MAX

    for i in range(M):
        (cP[ord(pat[i])]) += 1
        (cT[ord(txt[i])]) += 1

    for i in range(M,N):
        if compare(cP,cT):
            print("Found at Index",(i-M))
            
        (cT[ord(txt[i])]) += 1
        (cT[ord(txt[i-M])]) -= 1

    if compare(cP,cT):
        print("Found at index",N-M)
```

```python
txt = "BACDGABCDA"
pat = "ABCD"       
search(pat, txt) 
```

```python
txt = "AAABABAA" 
pat = "AABA"
search(pat, txt) 
```

**4. Last Match**

Given two strings A and B, you need to find the last occurrence ( 1 based indexing) of string B in string A.<br>

Example 1:<br>
Input:<br>
A = abcdefghijklghifghsd<br>
B = ghi<br>
Output: 13 <br>

Example 2: <br>
Input:<br>
A = abdbccaeab<br>
B = abc<br>
Output:-1

```python
def last_match(A, B):
    index = A.rfind(B)  
    if index == -1:
        return -1
    else:
        return index + 1 
```

```python
A = "abcdefghijklghifghsd"
B = "ghi"
print(last_match(A, B))
```

```python
A = "abdbccaeab"
B = "abc"
print(last_match(A, B))
```

**5. Pattern searching**

Given a string txt and a pattern pat. Your task is to return true if the pattern is present in the given string otherwise return false.<br>

Examples:<br>

Input: txt = "abcdefh", pat = "bcd"<br>
Output: true<br>

Input: txt = "axzy", pat = "xy"<br>
Output: false

```python
def pattern_search(txt, pat):
    n = len(txt)
    m = len(pat)
    for i in range(n - m + 1):
        if txt[i:i+m] == pat:
            return True
    return False
```

```python
A = "abcdefh"
B = "bcd"
print(pattern_search(A, B))
```

```python
A = "axzy"
B = "xy"
print(pattern_search(A, B))
```
