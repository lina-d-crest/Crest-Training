**1. Check Nineteen and Five Occurrences**

Write a Python program to find a list of integers with exactly two occurrences of nineteen and at least three occurrences of five. Return True otherwise False.<br>
Input:<br>
[19,19,15,5,3,5,5,2]<br>
Output:<br>
True<br>
Input:<br>
[19,15,15,5,3,3,5,2]<br>
Output:<br>
False<br>
Input:<br>
[19,19,5,5,5,5,5]<br>
Output:<br>
True 

```python
l = [19, 19, 5, 5, 5, 5, 5]
o1 = 0
o2 = 0

for i in l:
  if i == 19:
    o1 +=1
  elif i == 5:
    o2 += 1
    
if o1==2 and o2==3:
  print("True") 
else:
  print("False")
```

**2. Fifth Element and List Length Check**

Write a Python program that accepts a list of integers and calculates the length and the fifth element. Return true if the length of the list is 8 and the fifth element occurs thrice in the said list.<br>
Input:<br>
[19, 19, 15, 5, 5, 5, 1, 2]<br>
Output:<br>
True<br>
Input:<br>
[19, 15, 5, 7, 5, 5, 2]<br>
Output:<br>
False<br>
Input:<br>
[11, 12, 14, 13, 14, 13, 15, 14]<br>
Output:<br>
True<br>
Input:<br>
[19, 15, 11, 7, 5, 6, 2]<br>
Output:<br>
False

```python
list = [19, 15, 11, 7, 5, 6, 2]
length = len(list)
o = 0
x = list[4]

for i in list:
  if i == x:
    o += 1
    
if length==8 and o ==3:
  print("True")
else:
  print("False")
```
