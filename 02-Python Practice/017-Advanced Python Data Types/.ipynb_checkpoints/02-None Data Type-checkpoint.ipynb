**1. Descending Order List Check**

Write a Python function that checks if a given list is sorted in descending order. Return None if the list is empty.

```python
def fun(l):
    if not l:
        return None

    sd = all(l[i] >= l[i+1] for i in range(len(l)-1))
    return sd
```

```python
def main(n):

    try:
        r = fun(n)

        if r is None:
            print("List is Empty.")
        elif r:
            print("List is sorted in descending Order.")
        else:
            print("List is not sorted in Descending Order.")
    except Exception as e:
        print("An error occured: ",e)
```

```python
if __name__ == "__main__":
    main([7, 7, 8, 2, 1])
```

```python
if __name__ == "__main__":
    main([5,4,3,2,1])
```

```python
if __name__ == "__main__":
    main([])
```
