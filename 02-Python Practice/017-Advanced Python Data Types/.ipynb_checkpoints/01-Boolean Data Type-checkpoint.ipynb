**1. Boolean Operations**

Write a Python function that takes two boolean values as input and returns the logical AND and OR of those values.

```python
def fun(x,y):
    a = x and y
    b = x or y

    return a,b
```

```python
def main():

    try:
        i1 = input("Input 1st Boolean Value: ").lower()
        i2 = input("Input 2nd Boolean Value: ").lower()

        if i1 == "true":
            b1 = True
        elif i1 == "false":
            b1 = False
        else:
            print("Invalid input for first boolean value.")
            return

        if i1 == "true":
            b2 = True
        elif i1 == "false":
            b2 = False
        else:
            print("Invalid input for first boolean value.")
            return

        a,b = fun(b1,b2)
        print(f"Logical AND: {a}")
        print(f"Logical OR: {b}")
    except:
        print("Invalid Input.Please input either 'True' or False'.")
```

```python
main()
```
