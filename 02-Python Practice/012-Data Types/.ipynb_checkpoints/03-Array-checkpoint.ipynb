**1. Find the First Duplicate Element in an Array**

Write a Python program to find the first duplicate element in a given array of integers. Return -1 if there are no such elements.

```python
def fun(n):
    s = set()
    d = -1

    for i in range(len(n)):
        if n[i] in s:
            return n[i]
        else:
            s.add(n[i])
    return d
```

```python
fun([1,2,3,4,5])
```

```python
fun([1,2,3,4,4,5,6])
```

```python
fun([1, 1, 2, 3, 3, 2, 2])
```
