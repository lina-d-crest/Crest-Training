**1. Convert JSON Data to Python Object**

Write a Python program to convert JSON data to Python object.

```python
import json
```

```python
obj =  '{ "Name":"Alice", "Class":"I", "Age":6 }'

o = json.loads(obj)
print(f"JSON Data : {o}")
```

```python
print("Name: ",o["Name"])
print("Class: ",o["Class"])
print("Age: ",o["Age"]) 
```

<br>**2. Convert JSON Encoded Data into Python Objects**

Write a Python program to convert JSON encoded data into Python objects.

```python
d =  '{"name": "David", "age": 6, "class": "I"}'
l =   '["Red", "Green", "Black"]'
s = '"Python Json"'
i = '1234'
f =  '21.34'
```

```python
pd=  json.loads(d)
pl = json.loads(l)
ps =  json.loads(s)
pi =   json.loads(i)
pf = json.loads(f)
```

```python
print("Python dictionary: ", pd)
print("Python list: ", pl)
print("Python string: ", ps)
print("Python integer: ", pi)
print("Python float: ", pf)
```
