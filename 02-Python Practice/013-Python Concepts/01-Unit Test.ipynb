**1. Unit Test for Sorted List in Ascending Order**

Write a Python unit test program to check if a list is sorted in ascending order.

```python
import unittest
```

```python
def is_sorted_ascending(lst):
    return all(lst[i] <= lst[i+1] for i in range(len(lst) - 1))
```

```python
def is_sorted_ascending(lst):
    return all(lst[i] <= lst[i+1] for i in range(len(lst) - 1))

class TestSortedList(unittest.TestCase):

    def test_sorted_list(self):
        self.assertTrue(is_sorted_ascending([1, 2, 3, 4, 5]))

    def test_unsorted_list(self):
        self.assertFalse(is_sorted_ascending([5, 3, 4, 1]))

    def test_empty_list(self):
        self.assertTrue(is_sorted_ascending([]))

    def test_single_element(self):
        self.assertTrue(is_sorted_ascending([10]))
```

```python
if __name__ == '__main__':
    unittest.main(argv=[''], exit=False)
```
