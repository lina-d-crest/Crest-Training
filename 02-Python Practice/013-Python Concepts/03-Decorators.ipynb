**1. Create a Decorator to Log Function Arguments and Return Value**

Write a Python program to create a decorator that logs the arguments and return value of a function.

The decorator in this code logs the function name, arguments, and return value whenever the decorated function is called.

```python
def decorator(func):
    def wrap(*args,**kwargs):
        print(f"Calling {func.__name__} with args: {args},kwargs: {kwargs}")

        result = func(*args,**kwargs)
        
        print(f"{func.__name__} returned: {result}")

        return result
    return wrap
```

```python
@decorator
def multiply(x,y):
    return x*y
```

```python
result = multiply(10,20)
print("Result: ",result)
```

<br>**2. Create a decorator to convert function return value type**

Write a Python program to create a decorator to convert the return value of a function to a specified data type.

```python
def convert(data):
    def decorator(func):
        def wrapper(*args,**kwargs):

            result = func(*args,**kwargs)
            return data(result)
        return wrapper
    return decorator
```

```python
@convert(int)
def add(x,y):
    return x+y
```

```python
result = add(10,20)
print("Result: ",result,type(result))
```

```python
@convert(str)
def str(x,y):
    return x+y
```

```python
result = str("Python", " Decorator")
print("Result: ",result,type(result))
```
