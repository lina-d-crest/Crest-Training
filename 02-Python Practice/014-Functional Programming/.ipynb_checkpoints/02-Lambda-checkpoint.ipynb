**1. Next Bigger Number Lambda**

Write a Python program to create the next bigger number by rearranging the digits of a given number.

```python
def arrange(n):
    
    num = list(str(n))

    for i in range(len(num)-2,-1,-1):
        if num[i] < num[i+1]:

            z = num[i:]
            y = min(filter(lambda x: x>z[0],z))
            z.remove(y)
            z.sort()
            num[i:] = [y] + z

            return int("".join(num))
    return False
```

```python
n = 12
print("Original Number: ",n)
print("Next bigger Number: ",arrange(n))
```

```python
n = 10
print("Original Number: ",n)
print("Next bigger Number: ",arrange(n))
```
