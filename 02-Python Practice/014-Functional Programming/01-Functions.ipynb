**1. Maximum of Three Numbers**

Write a Python function to find the maximum of three numbers.

```python
def max2(x,y):
    if x>y:
        return x
    else:
        return y
```

```python
def max3(x,y,z):
    return max2(max2(x,y),z)
```

```python
x = int(input("x: "))
y = int(input("y: "))
z = int(input("z: "))

print("\nMaximum of 3 is: ",max3(x,y,z))
```

<br>**2. Invoke a Function After a Specified Period of Time**

Write a Python program that invokes a function after a specified period of time.

```python
from time import sleep
import math
```

```python
def delay(fn,ms,*args):
    sleep(ms/1000)
    return fn(*args)
```

```python
print("Square root after specific millisecond: ")
print(delay(lambda x: math.sqrt(x),100,16))
print(delay(lambda x: math.sqrt(x),1000,100))
print(delay(lambda x: math.sqrt(x),2000,25100))
```
